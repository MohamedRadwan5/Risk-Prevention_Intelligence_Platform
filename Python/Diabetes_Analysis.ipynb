**Outliers**

```python
pd.set_option('display.float_format', '{:,.2f}'.format)

country = df.drop_duplicates(['country_code', 'year'])
cols = ['diabetes_rate_age_standardized_pct', 'obesity_rate_pct', 'sugar_starch_pct_of_calories', 'gdp_per_capita_ppp',
        'health_spending_total_usd', 'health_coverage_index', 'country_diabetes_deaths', 'insulin_imports_usd', 'insulin_exports_usd']

rows = []
for c in cols:
    s = country[c].dropna().astype(float)
    q1, q3 = s.quantile([0.25, 0.75])
    low, high = q1 - 1.5 * (q3 - q1), q3 + 1.5 * (q3 - q1)
    out = s[(s < low) | (s > high)]
    rows.append({'column': c, 'outliers': len(out), 'pct': len(out) / len(s) * 100, 'min': s.min(), 'max': s.max(), 'upper_limit': high})

outliers_table = pd.DataFrame(rows)
outliers_table = outliers_table[outliers_table['outliers'] > 0].reset_index(drop=True)
outliers_table

def short(x):
    for size, unit in [(1e12, 'T'), (1e9, 'B'), (1e6, 'M'), (1e3, 'K')]:
        if abs(x) >= size:
            return f'{x / size:,.1f}{unit}'
    return f'{x:,.2f}'

display_table = outliers_table.copy()
for c in ['min', 'max', 'upper_limit']:
    display_table[c] = display_table[c].map(short)
display_table['pct'] = display_table['pct'].round(1)
display_table
```

**analyses**

Age-standardized diabetes prevalence among adults **18**

```python
# Here I will measure the prevalence rates among adults older than 18 years
# And I got a percentage that says out of every 100 adults how many get the disease diabetes_rate_age_standardized_pct
p = df[df['year'].between(2018, 2022)].drop_duplicates(['country', 'year'])

prevalence = p.groupby('country')[['diabetes_rate_age_standardized_pct']].mean()
prevalence['rank'] = prevalence['diabetes_rate_age_standardized_pct'].rank(ascending=False, method='min').astype(int)

level_labels = ['Low', 'Below average', 'Average', 'Above average', 'High']
prevalence['prevalence_level'] = pd.qcut(prevalence['diabetes_rate_age_standardized_pct'], 5, labels=level_labels)

prevalence = prevalence.sort_values('diabetes_rate_age_standardized_pct', ascending=False).round(2)

top10 = prevalence.head(10)
low10 = prevalence.tail(10).sort_values('diabetes_rate_age_standardized_pct')
egypt = prevalence.loc[['Egypt']].copy()
egypt['world_median'] = prevalence['diabetes_rate_age_standardized_pct'].median()

print(len(prevalence), 'countries')
display(top10.style.set_caption('Top 10 countries: age-standardized diabetes prevalence % (2018-2022)').format(precision=2))
display(low10.style.set_caption('Lowest 10 countries: age-standardized diabetes prevalence % (2018-2022)').format(precision=2))
display(egypt.style.set_caption('Egypt vs world median: diabetes prevalence (2018-2022)').format(precision=2))
```

I will calculate the causes of the high prevalence rate **bold text**

```python
# I measure the correlation of obesity and income and diet with the diabetes prevalence rate for each country from 2018 to 2022
# And I divide the countries into 4 groups by each factor and we compare the diabetes rate between them
# And I got a median for each group to measure the prevalence rate against

causes = p.groupby('country')[['obesity_rate_pct', 'gdp_per_capita_ppp', 'sugar_starch_pct_of_calories']].mean()

diet_lag = (df[df['year'].between(2008, 2012)].drop_duplicates(['country', 'year'])
            .groupby('country')['sugar_starch_pct_of_calories'].mean().rename('diet_10_years_before'))

causes = prevalence[['diabetes_rate_age_standardized_pct']].join(causes).join(diet_lag)

factors = ['obesity_rate_pct', 'gdp_per_capita_ppp', 'sugar_starch_pct_of_calories', 'diet_10_years_before']

corr_table = (causes[['diabetes_rate_age_standardized_pct'] + factors].corr(method='spearman')['diabetes_rate_age_standardized_pct']
              .drop('diabetes_rate_age_standardized_pct').round(2).rename('correlation').rename_axis('factor').reset_index())
corr_table['countries'] = [causes[f].notna().sum() for f in factors]

labels = ['Lowest', 'Below average', 'Above average', 'Highest']
for f in factors:
    causes[f + '_group'] = pd.qcut(causes[f], 4, labels=labels)

def by_group(col):
    t = causes.groupby(col, observed=True)['diabetes_rate_age_standardized_pct'].agg(countries='count', median_prevalence='median')
    t = t.rename_axis('group').reset_index()
    t['group'] = t['group'].astype(str)
    return t

groups = [('obesity_rate_pct_group', 'Obesity'),
          ('gdp_per_capita_ppp_group', 'Income per person'),
          ('sugar_starch_pct_of_calories_group', 'Sugar & starch diet'),
          ('diet_10_years_before_group', 'Diet 10 years before')]

groups_table = pd.concat([by_group(col).assign(factor=name) for col, name in groups],
                         ignore_index=True)[['factor', 'group', 'countries', 'median_prevalence']]

display(corr_table.style.set_caption('Correlation with diabetes prevalence (2018-2022)').format(precision=2))
display(groups_table.style.set_caption('Median diabetes prevalence by factor group (2018-2022)').format(precision=2)) # Here I took the median because some of my countries have outliers
# Like Micronesia the prevalence rate there is about 33%
```

Here I made an insight for Egypt to see and know where we stand in each group **bold text**

```python
#Here I found that Egypt is in the highest quarter in the world in obesity and starches and its income is in the middle and these are the same characteristics of the countries highest in diabetes so prevention must start from obesity and food
# And I found that 42 % of adults have obesity
names = ['Obesity', 'Income per person', 'Sugar & starch diet', 'Diet 10 years before']

egypt_rows = []
for f, name in zip(factors, names):
    egypt_rows.append({
        'factor': name,
        'egypt_value': causes.loc['Egypt', f],
        'egypt_group': causes.loc['Egypt', f + '_group'],
        'rank': int(causes[f].rank(ascending=False, method='min')['Egypt']),
        'countries': causes[f].notna().sum(),
        'world_median': causes[f].median(),
    })

egypt_table = pd.DataFrame(egypt_rows)
display(egypt_table.style.set_caption('Egypt: position in each factor group (2018-2022)').format(precision=2))
```

```python
# The total of patients here is calculated for each year so the patient who exists in every year is counted more than once
# That is why the number of patients in India is larger than its population so the percentage is calculated on the average number of patients who die each year

two = df[df['country'].isin(['India', 'Nauru']) & df['year'].between(2000, 2021)]

patients = two.groupby(['country', 'year'], as_index=False)['diabetes_patients'].sum()
deaths = two.drop_duplicates(['country', 'year'])[['country', 'year', 'country_diabetes_deaths']]
result = patients.merge(deaths, on=['country', 'year'])

totals = result.groupby('country')[['diabetes_patients', 'country_diabetes_deaths']].sum()
totals['deaths_pct'] = totals['country_diabetes_deaths'] / totals['diabetes_patients'] * 100
totals

# I found here that India has the largest number of diabetes deaths in the world yearly and Nauru the lowest yearly
# We will take into account the population difference and we will get the deaths percentage from the patients percentage yearly
```

Here I need to see the health care level for the two countries to see whether this has an effect on the deaths rate for the two countries or not **bold text**

```python
two = df[df['country'].isin(['India', 'Nauru']) & (df['year'] == 2019)]

two_care = two.drop_duplicates('country').set_index('country')[['health_coverage_index', 'healthcare_level', 'health_spending_total_usd', 'country_population', 'country_diabetes_deaths']]
patients = two.groupby('country')['diabetes_patients'].sum()
over30 = two[two['untreated_status'] == 'Known'].groupby('country')[['diabetes_patients', 'untreated_patients']].sum()

two_care['deaths_per_1000'] = two_care['country_diabetes_deaths'] / patients * 1000
two_care['spend_per_person'] = two_care['health_spending_total_usd'] / two_care['country_population']
two_care['treatment_gap_pct'] = (over30['untreated_patients'] / over30['diabetes_patients'] * 100).astype(float)

two_care[['deaths_per_1000', 'health_coverage_index', 'healthcare_level', 'spend_per_person', 'treatment_gap_pct']].round(2)
```

I will get the patients percentage and the deaths percentage for each country first and then see the causes that lead to death like the spending level and the treatment gap so I can determine whether they have a cause in the high death rates or not **bold text**

```python
# Here we took the sample on the years from 2015 to 2019 because from 2020 Corona came so there are many diabetes patients who died

y = df[df['year'].between(2015, 2019)]

patients = y.groupby(['country', 'year'])['diabetes_patients'].sum()
deaths = y.drop_duplicates(['country', 'year']).set_index(['country', 'year'])['country_diabetes_deaths'].astype(float)
yearly = pd.DataFrame({'diabetes_patients': patients, 'diabetes_deaths': deaths}).dropna()

problem = yearly.groupby('country')[['diabetes_patients', 'diabetes_deaths']].sum()
problem['deaths_per_1000'] = problem['diabetes_deaths'] / problem['diabetes_patients'] * 1000
problem['deaths_pct'] = problem['deaths_per_1000'] / 10
problem['rank'] = problem['deaths_per_1000'].rank(ascending=False, method='min').astype(int)

level_labels = ['Low', 'Below average', 'Average', 'Above average', 'High']
problem['deaths_level'] = pd.qcut(problem['deaths_per_1000'], 5, labels=level_labels)

problem = problem.sort_values('deaths_per_1000', ascending=False).round(2)

top10 = problem.head(10)
low10 = problem.tail(10).sort_values('deaths_per_1000')
egypt = problem.loc[['Egypt']].copy()
egypt['world_median'] = problem['deaths_per_1000'].median()

print(len(problem), 'countries')
display(top10.style.set_caption('Top 10 countries: diabetes deaths per 1,000 patients (2015-2019)').format(precision=2))
display(low10.style.set_caption('Lowest 10 countries: diabetes deaths per 1,000 patients (2015-2019)').format(precision=2))
display(egypt.style.set_caption('Egypt vs world median (2015-2019)').format(precision=2))
```

```python
# Here I prepared a table that has for each country the health coverage and the spending per person and the treatment gap from 2015 to 2019
# And I linked it to the deaths per 1000 patients to see what is related to the rise in deaths
country_years = y.drop_duplicates(['country', 'year']).copy()
country_years['spend_per_person'] = country_years['health_spending_total_usd'] / country_years['country_population']
care_info = country_years.groupby('country')[['health_coverage_index', 'spend_per_person']].mean()

over30 = y[y['untreated_status'] == 'Known'].groupby('country')[['diabetes_patients', 'untreated_patients']].sum()
care_info['treatment_gap_pct'] = (over30['untreated_patients'] / over30['diabetes_patients'] * 100).astype(float)

care_info['healthcare_level'] = pd.cut(care_info['health_coverage_index'], [-1, 59.999, 79.999, 101], labels=['Low', 'Medium', 'High'])

care = problem[['deaths_per_1000']].join(care_info)
care.shape
```

Here I will calculate the care level and the spending level and the treatment gap

```python
# We sort the countries and divide them into 4 equal quarters in number so I get the health spending levels and the treatment gap
labels = ['Lowest', 'Below average', 'Above average', 'Highest']
care['spend_group'] = pd.qcut(care['spend_per_person'], 4, labels=labels)
care['gap_group'] = pd.qcut(care['treatment_gap_pct'], 4, labels=labels)
```

```python
# Here I calculated the health care level for each country
# And I calculated the country spending per person
# And the treatment gap for patients older than 30 years who are not taking treatment
# And I made a table just for Egypt to see where Egypt is compared to the rest of the world countries
# And I made an index for universal health coverage for each country from 0 to 100  	health_coverage_index
def measure_tables(col, group_col, title):
    t = care[[col, group_col, 'deaths_per_1000']].dropna(subset=[col]).copy()
    t['rank'] = t[col].rank(ascending=False, method='min').astype(int)
    t = t.sort_values(col, ascending=False).round(2)

    top10 = t.head(10)
    low10 = t.tail(10).sort_values(col)
    egypt = t.loc[['Egypt']].copy()
    egypt['world_median'] = t[col].median()

    display(top10.style.set_caption(f'Top 10 countries: {title} (2015-2019)').format(precision=2))
    display(low10.style.set_caption(f'Lowest 10 countries: {title} (2015-2019)').format(precision=2))
    display(egypt.style.set_caption(f'Egypt vs world median: {title} (2015-2019)').format(precision=2))

measure_tables('health_coverage_index', 'healthcare_level', 'Health coverage index')
measure_tables('spend_per_person', 'spend_group', 'Health spending per person (USD)')
measure_tables('treatment_gap_pct', 'gap_group', 'Treatment gap (%)')
```

Causes of the treatment gap **bold text**

```python
# Here I see why there are countries where most of their patients are not being treated from 2015 to 2019
# I measure the correlation of the treatment gap with health coverage and spending per person and income per person and income level
# And I divided the countries into 4 groups by each factor and got the median treatment gap for each group and a table for Egypt

gdp = y.drop_duplicates(['country', 'year']).groupby('country')['gdp_per_capita_ppp'].mean()
income = y[y['year'] == 2019].drop_duplicates('country').set_index('country')['income_level']

gap_causes = care[['treatment_gap_pct', 'health_coverage_index', 'spend_per_person']].join(gdp).join(income)

gap_factors = ['health_coverage_index', 'spend_per_person', 'gdp_per_capita_ppp']

gap_corr = (gap_causes[['treatment_gap_pct'] + gap_factors].corr(method='spearman')['treatment_gap_pct']
            .drop('treatment_gap_pct').round(2).rename('correlation').rename_axis('factor').reset_index())
gap_corr['countries'] = [gap_causes[f].notna().sum() for f in gap_factors]

labels = ['Lowest', 'Below average', 'Above average', 'Highest']
for f in gap_factors:
    gap_causes[f + '_group'] = pd.qcut(gap_causes[f], 4, labels=labels)

def gap_by_group(col):
    t = gap_causes[gap_causes[col] != 'Unknown'].groupby(col, observed=True)['treatment_gap_pct'].agg(countries='count', median_gap='median')
    t = t.rename_axis('group').reset_index()
    t['group'] = t['group'].astype(str)
    return t

gap_groups = [('health_coverage_index_group', 'Health coverage'),
              ('spend_per_person_group', 'Health spending per person'),
              ('gdp_per_capita_ppp_group', 'Income per person'),
              ('income_level', 'Income level (World Bank)')]

gap_groups_table = pd.concat([gap_by_group(col).assign(factor=name) for col, name in gap_groups],
                             ignore_index=True)[['factor', 'group', 'countries', 'median_gap']]

egypt_rows = []
for f, name in zip(gap_factors, ['Health coverage', 'Health spending per person', 'Income per person']):
    egypt_rows.append({
        'factor': name,
        'egypt_value': gap_causes.loc['Egypt', f],
        'egypt_group': gap_causes.loc['Egypt', f + '_group'],
        'rank': int(gap_causes[f].rank(ascending=False, method='min')['Egypt']),
        'countries': gap_causes[f].notna().sum(),
        'world_median': gap_causes[f].median(),
    })
egypt_gap_table = pd.DataFrame(egypt_rows)

display(gap_corr.style.set_caption('Correlation with treatment gap (2015-2019)').format(precision=2))
display(gap_groups_table.style.set_caption('Median treatment gap % by factor group (2015-2019)').format(precision=2))
display(egypt_gap_table.style.set_caption('Egypt: position in each factor (2015-2019)').format(precision=2))

# I found here that there is a very strong inverse relationship between the treatment gap and health coverage
# And the country spending on each person and income per person
# And here something very important became clear to us which is the strongest relationship that explains the cause of the treatment gap
# Whenever health coverage increases the treatment gap decreases with it
```

Egypt priorities by **age**

```python
# Here I see in Egypt in 2022 which age group has the largest number of patients who are not being treated
# To determine from what age early screening starts (treatment data is available from 30 years only)

egypt_age = df[(df['country_code'] == 'EGY') & (df['year'] == 2022) & (df['untreated_status'] == 'Known')]

priority = egypt_age.groupby('age_group', observed=True)[['population', 'diabetes_patients', 'untreated_patients']].sum().reset_index()
priority['age_group'] = priority['age_group'].astype(str)
priority['prevalence_pct'] = priority['diabetes_patients'] / priority['population'] * 100
priority['untreated_pct'] = priority['untreated_patients'].astype(float) / priority['diabetes_patients'] * 100
priority['share_of_untreated'] = priority['untreated_patients'].astype(float) / priority['untreated_patients'].sum() * 100

focus = priority[priority['age_group'].isin(['30-34', '35-39', '40-44', '45-49', '50-54', '55-59'])]
summary = pd.DataFrame({
    'ages': ['30-59'],
    'untreated_patients': [int(focus['untreated_patients'].sum())],
    'share_of_all_untreated': [focus['share_of_untreated'].sum()],
})

sex_gap = egypt_age.groupby('sex', observed=True)[['diabetes_patients', 'untreated_patients']].sum()
sex_gap['untreated_pct'] = sex_gap['untreated_patients'].astype(float) / sex_gap['diabetes_patients'] * 100

def to_m(x):
    return f'{x / 1e6:,.2f}M'

def to_pct(x):
    return f'{x:.1f}%'

display(priority.style.set_caption('Egypt 2022: diabetes and untreated patients by age group (30+)')
        .format({'population': to_m, 'diabetes_patients': to_m, 'untreated_patients': to_m,
                 'prevalence_pct': to_pct, 'untreated_pct': to_pct, 'share_of_untreated': to_pct}))
display(summary.style.set_caption('Egypt 2022: untreated patients aged 30-59')
        .format({'untreated_patients': to_m, 'share_of_all_untreated': to_pct}))
display(sex_gap.style.set_caption('Egypt 2022: untreated patients by sex (30+)')
        .format({'diabetes_patients': to_m, 'untreated_patients': to_m, 'untreated_pct': to_pct}))
```

```python
# We found here that whenever the age group gets younger the treatment gap in that group gets bigger
# The recommendation here is that we must provide awareness campaigns or facilities for the patients of this group who are not taking treatment
# The group from age 30 to age 59
```
